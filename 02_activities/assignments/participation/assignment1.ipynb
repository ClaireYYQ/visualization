Recall the aesthetic, substantive, and perceptual qualities of data visualizations. Does your visualization of choice succeed in each area?

Aesthetic quality: Contour plot uses smooth contour lines and color gradients that create an attractive appearance.

Substance quality: The contour plot summarizes the relationship between two continuous variables by showing regions of higher and lower density. It reveals patterns that may be difficult to see in a scatterplot with many overlapping observations.

Perceptual quality: The contour plot is reasonably effective for identifying trends and clusters

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests

np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0, 100,50)
y2 = np.random.randint(0, 100,50)
```

```python
# modify legend
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc = 'lower right',
          frameon = True, #add frame around the legend
          fontsize = 12,
          ncol = 2, #specify number of columns
          shadow = True )
fig.show()
```

```python
# move legend ourside of the plot
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc = 'upper left',
          bbox_to_anchor = (1, 1))
fig.show()
```

```python
# annotations with text() and modify it
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc = 'lower right')
ax.text(10, 95, "This value is important!",
        ha ='center',# alignment of text
        color = 'red',  # modify font colour
        size = 20)      
fig.show()
```

```python
#position text annotations with transform
fig, ax = plt.subplots()
ax.axis([0, 10, 0, 10])
ax.text(1, 5, 
        ". Data:(1, 5)", 
        transform = ax.transData)       
ax.text(0.5, 0.1, 
        ". Axes:(0.5, 0.1)",
        transform = ax.transAxes)
ax.text(0.2, 0.2, 
        ". Figure:(0.2, 0.2)", 
        transform = fig.transFigure)
fig.show()
```

```python
# annotation with arrows
fig, ax = plt.subplots(figsize = (5, 3))
ax.scatter(x, y1, label = "Person 1")
ax.scatter(x, y2, label = "Person 2")
ax.legend(loc = 'lower right')
ax.annotate('This is important!', 
            xy = (10, 95), 
            xytext = (20, 94),
            arrowprops = dict(facecolor = 'black'))
fig.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0, 75,50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])

# Introduce subplots
fig,(ax1, ax2) = plt.subplots(ncols = 2, nrows = 1, figsize = (7, 3))
ax1.scatter(x1, y1)
ax2.bar(x2, y2)
fig.show()
```

```python
# customize the plot
fig,(ax1, ax2) = plt.subplots(ncols = 2, nrows = 1, figsize = (7, 3))
ax1.scatter(x1, y1, marker = '*', color = 'indigo')
ax2.bar(x2, y2, color = 'pink')
ax1.annotate('This is important!', 
            xy = (30, 70), 
            xytext = (20, 80),
            arrowprops = dict(facecolor = 'black'))
fig.show()
```

```python
# grid arrangement
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'], ['ax2', 'ax3']], figsize = (7, 4))
someaxes["ax1"].scatter(x1, y1)
someaxes["ax2"].bar(x2, y2)
someaxes["ax3"].plot(x1, y1)
plt.show()
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'], ['ax2', 'ax3']], figsize = (7, 4))
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
someaxes["ax1"].set_xlabel('A Big Label', fontsize = 18)
someaxes["ax2"].set_xlabel('Another Label', fontsize = 18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious', fontsize = 18)
fig.show()
```

```python
#Multiple viz on one axes object
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])

fig, ax = plt.subplots(figsize = (7, 3))
ax.bar(x, y1, color = "indigo")
ax.plot(x, y2, color = "red")
fig.show()
```

```python
# Add error information
y2_sd = np.std(y2)
fig, ax = plt.subplots(figsize = (7, 3))
ax.plot(x, y2, color = "red")
ax.errorbar(x, y2, yerr = y2_sd, fmt = "none", 
            ecolor = "indigo", elinewidth = 4, capsize = 6, capthick= 4, # customize error
            errorevery = 2) # specify intervals
```

```python
# Get image from internet
from PIL import Image # to open images
import requests # to get images from URLs
from io import BytesIO # to store images

url = "https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png"
headers = {"User-Agent": "Mozilla/5.0","Accept": "image/*,*/*;q=0.8","Referer": "https://en.wikipedia.org/"}
response = requests.get(url, headers=headers)
image_file = BytesIO(response.content)
image = Image.open(image_file)
image
```

```python
# Add image into plot
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")
ax_image = fig.add_axes([0.1, # x coordinate (ON FIGURE, NOT AXES)
                         0.11, # y coordinate (ON FIGURE, NOT AXES)
                         0.15, # image width
                         0.35]) # image height
ax_image.imshow(image)
ax_image.axis ('off')
```

```python
# Customize my own
from matplotlib.offsetbox import OffsetImage, AnnotationBbox
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")
for xi, yi in zip(x, y2): # pair data into coordinates
    im = OffsetImage (image, zoom=0.2)  # transform into small icon
    ab = AnnotationBbox(im, (xi, yi), frameon=False) # place image at data coordinates
    ax.add_artist(ab) # adds the image to the plot
ax_image.imshow(image)
ax_image.axis ('off')
```

```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
#This dataset 'tips' contains information about tips at a restaurant, including meal time, party size, gender of customer, total bill amount, and tip amount

tips = sns.load_dataset("tips")
print(tips)
```

```python
sns.lineplot(data = tips,
             x = 'total_bill',
             y = 'tip')

# premade styles: ‘darkgrid’,‘whitegrid’,‘dark’,‘white’,‘ticks’...
#sns.set_style('whitegrid')
sns.set_style('ticks')
```

```python
# Add labels using set. function
tipgraph = sns.lineplot(data = tips,
                        x = 'total_bill',
                        y = 'tip')
tipgraph.set(title = 'Tips vs. Total Bill',
             xlabel = 'Total Bill ($)',
             ylabel = 'Tip Amount ($)')
```

```python
# Use the subplots function to change figure size
fig = plt.subplots(figsize = (10, 3))

# Modify aesthetic elements
tipgraph = sns.lineplot(data = tips,
                        x = 'total_bill',
                        y = 'tip',
                        color = 'hotpink',
                        linestyle = '--',
                        linewidth = 3,
                        marker = 'o',
                        markerfacecolor = 'indigo')
```

```python
tipgraph = sns.scatterplot(data = tips, 
                           x = 'total_bill',
                           y = 'tip',
                           style = 'time', 
                           hue = 'day', 
                           palette = ['purple','hotpink','deepskyblue','yellowgreen'])
tipgraph.set(title = 'Tips vs. Total Bill',
             xlabel = 'Total Bill ($)',
             ylabel = 'Tip Amount ($)')
```

```python
# Pairplot
sns.pairplot(data = tips, hue = 'day')
```

```python
# Relplot, a high-level interface for plotting relationships between variables
daysplt = sns.relplot(data = tips,
                      x = 'total_bill',
                      y = 'tip',
                      hue = 'sex', # add color grouping (categorical separation)
                      col = 'day', # separate subplots for each category in day.
                      kind = 'scatter', # specify scatter plot for each panel
                      col_wrap = 2) # layout of the subplots, row = 2
```

```python
# !pip install wordcloud
from wordcloud import WordCloud
df = pd.read_csv("https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/movie_quotes.csv",
                 on_bad_lines='skip')
df
```

```python
# join all our text from each row from our quote column into a string
text = " ".join(each for each in df.quote)

# generate our wordcloud image
wordcloud = WordCloud(background_color = "white",
colormap = 'inferno').generate(text)

# use matplotlib syntax to put our image in a figure
fig, ax = plt.subplots(figsize=(7, 3))
ax.imshow(wordcloud, # remember 'imshow' from when we added pictures to our matplotlib axes
          interpolation='bilinear') # this line helps smooth our image
ax.axis("off")
```

```python
#!pip install matplotlib-venn
from matplotlib_venn import venn2, venn2_circles, venn2_unweighted

A = set(["apple","banana","watermelon"])
B = set(["pumpkin","blueberry","apple","key lime"])
diagram = venn2_unweighted([A, B],
                           set_labels = ('Fruits','Pies'),
                           set_colors = ('blue','red'),
                           alpha = 0.5)

diagram.get_label_by_id("10").set_text("\n".join(A - B))
diagram.get_label_by_id("11").set_text("\n".join(A & B))
diagram.get_label_by_id("01").set_text("\n".join(B - A))

plt.show()
```
